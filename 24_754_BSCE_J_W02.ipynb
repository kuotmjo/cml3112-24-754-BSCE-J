#### 1. What happens when you repeat a cell?
Repeating a cell executes the code block again using the current state of the environment's memory. If a cell modifies or appends data to a variable, repeating it will continually alter that variable, increment the execution number counter, and potentially cause unexpected errors if the initial variables are not reset first.

#### 2. What does "Restart and Run All" do?
The "Restart and Run All" function completely clears the notebook's active memory kernel, resets all variables, and executes every single cell sequentially from the very top to the bottom. This ensures the notebook runs cleanly from scratch without relying on hidden, out-of-order execution states.
